## Jak czytać komórkę sprawdzenia

Pod każdym zadaniem jest komórka sprawdzenia. Składa się z linii `assert`. Taka linia ma dwie części: warunek i komunikat. Pierwsza linia sprawdzenia A2.1 wygląda tak:

```python
assert remaining == 17, "A2.1: oblicz stan po dostawie i sprzedaży; wynik ma być liczbą."
```

Python liczy warunek `remaining == 17`. Gdy jest prawdziwy, przechodzi do następnej linii bez żadnego napisu. Gdy jest fałszywy, zatrzymuje komórkę i wyświetla komunikat. Sprawdzenie porównuje więc Twoje wartości z opisem zadania, linia po linii, i zatrzymuje się na pierwszej, która nie pasuje.

Jeżeli uruchomisz sprawdzenie A2.1 przed rozwiązaniem zadania, zobaczysz czerwony blok podobny do tego:

```text
AssertionError                            Traceback (most recent call last)
Cell In[3], line 1
----> 1 assert remaining == 17, "A2.1: oblicz stan po dostawie i sprzedaży; wynik ma być liczbą."
      2 assert label == "known", "A2.1: label ma być napisem known."
      3 print("A2.1: wyniki pasują do opisu.")

AssertionError: A2.1: oblicz stan po dostawie i sprzedaży; wynik ma być liczbą.
```

Czytaj go od końca:

- ostatnia linia zaczyna się od `AssertionError:` i zawiera komunikat; to jest wskazówka, co poprawić,
- strzałka `---->` pokazuje, która linia sprawdzenia się zatrzymała; linie pod nią jeszcze się nie wykonały,
- `AssertionError` oznacza niespełniony warunek sprawdzenia. Inna nazwa błędu oznacza, że Python nie zdołał wykonać sprawdzenia: przeczytaj ostatnią linię i sprawdź wskazane miejsce. Przy `NameError` upewnij się najpierw, że komórka zadania została uruchomiona.

Gdy wszystkie warunki są spełnione, komórka wypisuje jedną linię, np. `A2.1: wyniki pasują do opisu.`

Po każdej poprawce uruchom najpierw komórkę zadania, a dopiero potem sprawdzenie. Sprawdzenie czyta wartości z pamięci kernela, więc sama zmiana tekstu w komórce zadania nic nie zmienia, dopóki jej nie uruchomisz.
