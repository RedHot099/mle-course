## B6.1 · Napisz funkcję w pliku

W edytorze otwórz `practice/summary.py`, obok folderu labs. Uzupełnij `mean_known(values)` według docstringa. Przyjmuje listę nieujemnych liczb i None. Zwraca średnią znanych wartości jako float, uwzględnia zera i nie zmienia listy. Dla pustej listy lub samych None zgłasza ValueError z komunikatem `no known values`.

Nie obsługuj tu plików i nie wypisuj wyniku z funkcji. Nie musisz sprawdzać innych typów ani liczb ujemnych.

Po zapisaniu pliku wykonaj gotową komórkę poniżej. Wczytuje plik ponownie, aby sprawdzenie korzystało z zapisanej wersji. Samą składnię importu poznasz na wykładzie B7.1.

````python
namespace = runpy.run_path(str(module_path))
mean_known = namespace["mean_known"]
````

````python
import math
values = [0, None, 8]
result = mean_known(values)
assert result == 4.0 and isinstance(result, float), "B6.1: uwzględnij zero i zwróć float, nie print."
assert values == [0, None, 8], "B6.1: nie zmieniaj listy."
assert math.isclose(mean_known([1, 2, 2]), 5 / 3), "B6.1: nie zaokrąglaj wyniku po drodze."
assert mean_known([0, 0]) == 0.0, "B6.1: znane zera mają średnią 0."
for values in [[], [None, None]]:
    try:
        mean_known(values)
    except ValueError as error:
        assert "no known values" in str(error), "B6.1: podaj przyczynę braku średniej."
    else:
        raise AssertionError("B6.1: bez pomiarów potrzebny jest ValueError.")
print("B6.1: funkcja spełnia sprawdzone przypadki umowy.")
````
