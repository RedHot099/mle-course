## B7.2 · Gotowe CI i opcjonalny pull request

Skopiuj dostarczony `lectures/resources/checks.yml` do `.github/workflows/checks.yml` w swojej kopii repozytorium. Nie nadpisuj istniejącego workflow bez porównania. Ten plik uruchamia narzędzia tylko w practice, nie sprawdza wszystkich nierozwiązanych notebooków.

W terminalu z głównego folderu repozytorium:

```bash
mkdir -p .github/workflows
cp lectures/resources/checks.yml .github/workflows/checks.yml
git add .github/workflows/checks.yml
git diff --cached
git commit -m "Check observation summary in CI"
git remote -v
```

Dalszy krok wykonaj, gdy origin wskazuje Twoje repozytorium na GitHub i chcesz wysłać swoją gałąź:

```bash
git push -u origin learning/summary
```

Na GitHub otwórz pull request do gałęzi bazowej własnego repozytorium. W Checks znajdź formatowanie, lint, typy i testy. Jeżeli nie masz konta lub Actions są niedostępne, ukończ lokalne sprawdzenia i przejdź do C. Dostępu do usługi nie traktujemy jako sprawdzianu Pythona.
